```python
# tsak 3.2
import pandas as pd

employee_data = pd.DataFrame([
    {'name': 'David', 'salary': 50000, 'department': 'Sales'},
    {'name': 'Eve', 'salary': 60000, 'department': 'Marketing'},
    {'name': 'Frank', 'salary': 55000, 'department': 'Sales'}
])

print(employee_data)
```
